```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
turnout_by_province = df.groupby('province')['turnout'].mean().sort_values(ascending=False)

plt.figure(figsize=(8,5))
sns.barplot(x=turnout_by_province.index, y=turnout_by_province.values)
plt.title('Average Voter Turnout by Province (2018)')
plt.ylabel('Turnout')
plt.xlabel('Province')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
party_seats = df['win_party'].value_counts().head(6)

plt.figure(figsize=(8,5))
sns.barplot(x=party_seats.values, y=party_seats.index)
plt.title('Seats Won by Party (Top 6)')
plt.xlabel('Seats')
plt.tight_layout()
plt.show()
```

```python
df['margin'] = df['win_pct'] - df['second_pct']
df.groupby('province')['margin'].mean().sort_values()
```

```python
df[['turnout','margin']].corr()
```

## Statistical Analysis
- Winning margin (win_pct - second_pct) varies by province: Punjab is most 
  competitive (~10% margin), Islamabad least (~23% margin)
- Turnout and margin show a weak negative correlation (-0.20), suggesting 
  higher turnout is mildly associated with closer, more competitive races

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='turnout', y='margin', hue='province')
plt.title('Turnout vs Winning Margin by Province')
plt.xlabel('Turnout')
plt.ylabel('Winning Margin')
plt.tight_layout()
plt.show()
```

## Insights
1. Voter turnout varies notably by region: Islamabad had the highest turnout 
   (~58%), Khyber Pakhtunkhwa the lowest (~44%).
2. Pakistan Tehreek-e-Insaf (PTI) won the most seats nationally (331), followed 
   by PMLN (200) and PPP (128), reflecting its strong performance in the 2018 
   election.
3. Punjab had the most competitive races (smallest average winning margin), 
   while Islamabad's races were the least competitive.
4. Higher turnout is weakly associated with closer races (correlation -0.20), 
   suggesting competitive constituencies may drive slightly higher engagement.

## Limitations
- This dataset covers only the 2018 election; it doesn't show shifts across 
  multiple election cycles.
- Demographic data (income, urban/rural, age) is not included, limiting deeper 
  demographic analysis.
- Results are provisional/as-reported and may contain minor ECP data entry 
  discrepancies.
